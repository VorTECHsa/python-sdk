### Try out the VortexaSDK

First let's import our requirements

```python
from vortexasdk import VoyageCalculator
```

You'll need to enter your Vortexa API key when prompted.

```python
# Calculate ETA for a VLCC travelling from Ras Tanura to Rotterdam at 12 knots
ras_tanura = "539db1548407fd97024391d01a6a2be239b1100f070a137d54a79907d03db6c8"
rotterdam = "68faf65af1345067f11dc6723b8da32f00e304a6f33c000118fccd81947deb4e"

result = VoyageCalculator().search(
    type="ETA",
    vessel_status="vessel_status_laden_known",
    origin=ras_tanura,
    destination=rotterdam,
    vessel_class="oil_vlcc",
    ETD="2024-03-01T00:00:00.000Z",
    speed=12,
)
df = result.to_df()
```

```python
df.head()
```

```python
# Calculate speed required to arrive by a specific date
speed_result = VoyageCalculator().search(
    type="speed",
    vessel_status="vessel_status_laden_known",
    origin=ras_tanura,
    destination=rotterdam,
    vessel_class="oil_vlcc",
    ETD="2024-03-01T00:00:00.000Z",
    ETA="2024-04-01T00:00:00.000Z",
)
speed_result.to_df()
```

```python
# Use lat/long coordinates as origin and avoid the Suez Canal
latlong_result = VoyageCalculator().search(
    type="ETA",
    vessel_status="vessel_status_laden_known",
    origin={"lat": 26.6, "long": 50.1},
    destination=rotterdam,
    vessel_class="oil_vlcc",
    ETD="2024-03-01T00:00:00.000Z",
    speed=12,
    avoid_zone=["Suez Canal"],
)
latlong_result.to_df()
```
